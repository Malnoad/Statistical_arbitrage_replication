```python
# Vẽ lại chart tổng hợp nếu cần
import matplotlib.pyplot as plt
import matplotlib.dates as mdates

fig, axes = plt.subplots(2, 1, figsize=(14, 12))

# Equity curves
ax = axes[0]
ax.plot(equity_fixed, label='Fixed (15 Factors)',     color='blue')
ax.plot(equity_var,   label='Variable (55% Variance)', color='red')
ax.plot(equity_syn,   label='Synthetic ETFs',          color='green')
ax.plot(equity_act,   label='Actual ETFs',             color='orange')
ax.set_title('Equity Curves — 4 Strategies')
ax.set_ylabel('Equity Value')
ax.legend()
ax.grid(True, alpha=0.3)

# PnL curves
ax = axes[1]
ax.plot(equity_fixed - 100, label='Fixed (15 Factors)',     color='blue')
ax.plot(equity_var   - 100, label='Variable (55% Variance)', color='red')
ax.plot(equity_syn   - 100, label='Synthetic ETFs',          color='green')
ax.plot(equity_act   - 100, label='Actual ETFs',             color='orange')
ax.axhline(y=0, color='black', linestyle='--', alpha=0.5)
ax.set_title('PnL Curves — 4 Strategies')
ax.set_ylabel('Cumulative PnL')
ax.legend()
ax.grid(True, alpha=0.3)

for ax in axes:
    ax.xaxis.set_major_locator(mdates.YearLocator())
    ax.xaxis.set_major_formatter(mdates.DateFormatter('%Y'))

plt.tight_layout()
plt.show()
```
